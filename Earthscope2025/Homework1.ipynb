# 2. How many documents are in the site collection with station latitude greater than or equal to 45.0 degrees?

```python
import pandas as pd
def print_as_table(doclist):
    df = pd.DataFrame.from_dict(doclist)
    display(df)
```

```python
query={'lat' : {'$gte' : 45.0}}

cursor=db.site.find(query)
doclist=[]
for doc in cursor:
    doclist.append(doc)
print_as_table(doclist)
```
